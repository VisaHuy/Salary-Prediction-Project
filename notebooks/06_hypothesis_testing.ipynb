```python
import pandas as pd
import numpy as np
from scipy import stats
from scipy.stats import norm

alpha = np.float64(0.05)
```

```python
# Load cleaned dataset
df = pd.read_csv('../data/data_cleaned.csv')
df = df.sample(n=300, random_state=42)
```

### Question 1

**Question:** Do people who earn `>50K` work a different number of hours per week on average compared with people who earn `<=50K`?

- **H0**: The two income groups have the same average `hours-per-week`.
- **H1**: The two income groups have different average `hours-per-week`.

```python
group_high = df[df['income'] == '>50K']['hours-per-week']
group_low = df[df['income'] == '<=50K']['hours-per-week']

result_1 = stats.ttest_ind(group_high, group_low)
p_val_1 = result_1.pvalue
mean_diff_1 = group_high.mean() - group_low.mean()
decision_1 = 'Reject H0' if p_val_1 < alpha else 'Fail to reject H0'
conclusion_1 = (
    'The average working hours are significantly different between the two income groups.'
    if p_val_1 < alpha
    else 'There is not enough evidence to say the average working hours are different.'
)

print('Question 1 Results')
print(f'Average hours (>50K): {group_high.mean():.3f}')
print(f'Average hours (<=50K): {group_low.mean():.3f}')
print(f'Difference in averages (>50K - <=50K): {mean_diff_1:.3f}')
print(f'p-value: {p_val_1:.6g}')
print('Decision:', decision_1)
print('Conclusion:', conclusion_1)
```

## Question 2

**Question:** Is the average `education-num` different between people earning `>50K` and `<=50K`?

- **H0**: Both income groups have the same average `education-num`.
- **H1**: The two income groups have different average `education-num`.

```python
group_high_edu = df[df['income'] == '>50K']['education-num']
group_low_edu = df[df['income'] == '<=50K']['education-num']

result_2 = stats.ttest_ind(group_high_edu, group_low_edu, equal_var=False)
p_val_2 = result_2.pvalue
mean_diff_2 = group_high_edu.mean() - group_low_edu.mean()
decision_2 = 'Reject H0' if p_val_2 < alpha else 'Fail to reject H0'
conclusion_2 = (
    'The average education-num is significantly different between the two income groups.'
    if p_val_2 < alpha
    else 'There is not enough evidence to say the average education-num is different.'
)

print('Question 2 Results')
print(f'Average education-num (>50K): {group_high_edu.mean():.3f}')
print(f'Average education-num (<=50K): {group_low_edu.mean():.3f}')
print(f'Difference in averages (>50K - <=50K): {mean_diff_2:.3f}')
print(f'p-value: {p_val_2}')
print('Decision:', decision_2)
print('Conclusion:', conclusion_2)
```

## Question 3

**Question:** Is the proportion of people earning `>50K` different between males and females?

- **H0**: Males and females have the same proportion of people earning `>50K`.
- **H1**: Males and females have different proportions of people earning `>50K`.

```python
male = df[df['sex'] == 'Male']
female = df[df['sex'] == 'Female']

x1 = (male['income'] == '>50K').sum()
n1 = len(male)
x2 = (female['income'] == '>50K').sum()
n2 = len(female)

p1 = x1 / n1
p2 = x2 / n2
p_pool = (x1 + x2) / (n1 + n2)
diff_prop = p1 - p2

se_pool = np.sqrt(p_pool * (1 - p_pool) * (1 / n1 + 1 / n2))
z_stat = diff_prop / se_pool
p_val_3 = 2 * (1 - norm.cdf(abs(z_stat)))
decision_3 = 'Reject H0' if p_val_3 < alpha else 'Fail to reject H0'
conclusion_3 = (
    'The proportion of people earning >50K is significantly different between males and females.'
    if p_val_3 < alpha
    else 'There is not enough evidence to say the two proportions are different.'
)

print('Question 3 Results')
print(f'High-income proportion (Male): {p1:.4f} ({x1}/{n1})')
print(f'High-income proportion (Female): {p2:.4f} ({x2}/{n2})')
print(f'Difference in proportions (Male - Female): {diff_prop:.4f}')
print(f'p-value: {p_val_3:.6g}')
print('Decision:', decision_3)
print('Conclusion:', conclusion_3)
```
