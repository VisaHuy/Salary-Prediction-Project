# DATA CLEANING PROCESS

## Step 1: Handle Missing Values

```python
# Replace '?' with NaN for proper handling
df = df.replace('?', np.nan)
df = df.replace(' ?', np.nan)

# Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum())
print(f"\nPercentage of missing values:")
missing_percent = (df.isnull().sum() / len(df)) * 100
print(missing_percent[missing_percent > 0].round(2))
```

```python
# Strategy: For categorical variables with missing values, we'll use mode imputation
# or drop rows if missing percentage is very low

# Check which columns have missing values
cols_with_missing = df.columns[df.isnull().any()].tolist()
print(f"Columns with missing values: {cols_with_missing}")

# For this dataset, we'll drop rows with missing values since the percentage is typically low (<5%)
rows_before = len(df)
df = df.dropna()
rows_after = len(df)
print(f"\nRows removed: {rows_before - rows_after} ({((rows_before - rows_after)/rows_before)*100:.2f}%)")
print(f"Remaining rows: {rows_after}")
```

## Step 2: Remove Duplicates

```python
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Duplicates removed. New shape: {df.shape}")
else:
    print("No duplicates found.")
```

## Step 3: Clean Text Data (Strip Whitespace)

```python
categorical_cols = df.select_dtypes(include=['object']).columns

print("Cleaning text data (removing leading/trailing whitespaces and dots)...")
for col in categorical_cols:
    df[col] = df[col].str.strip()
    df[col] = df[col].str.rstrip('.')
    
print("Text data cleaned successfully!")
print("\nSample values after cleaning:")
print(df[categorical_cols].head())
print("\nUnique income values:", df['income'].unique())
```

## Step 4: Validate Data Ranges and Business Logic

```python
print("Checking data validity...")

invalid_age = df[(df['age'] < 17) | (df['age'] > 90)]
print(f"Invalid age records: {len(invalid_age)}")

invalid_hours = df[(df['hours-per-week'] < 1) | (df['hours-per-week'] > 100)]
print(f"Invalid hours-per-week records: {len(invalid_hours)}")

invalid_capital = df[(df['capital-gain'] < 0) | (df['capital-loss'] < 0)]
print(f"Invalid capital records: {len(invalid_capital)}")

invalid_edu = df[(df['education-num'] < 1) | (df['education-num'] > 16)]
print(f"Invalid education-num records: {len(invalid_edu)}")

initial_rows = len(df)
df = df[(df['age'] >= 17) & (df['age'] <= 90)]
df = df[(df['hours-per-week'] >= 1) & (df['hours-per-week'] <= 100)]
df = df[(df['capital-gain'] >= 0) & (df['capital-loss'] >= 0)]
df = df[(df['education-num'] >= 1) & (df['education-num'] <= 16)]
print(f"\nRows removed due to invalid values: {initial_rows - len(df)}")
print(f"Remaining rows: {len(df)}")
```

## Step 5: Handle Outliers (IQR Method)

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
print(f"Numerical columns: {numerical_cols}")

# Calculate outliers using IQR method
def detect_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    return len(outliers), lower_bound, upper_bound

print("\nOutlier Analysis:")# Identify numerical columns for outlier detection

print("-" * 70)
for col in numerical_cols:
    n_outliers, lower, upper = detect_outliers_iqr(df, col)
    print(f"{col}: {n_outliers} outliers (Valid range: {lower:.2f} to {upper:.2f})")
```

```python
# Visualize outliers with boxplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()

for idx, col in enumerate(numerical_cols):
    if idx < len(axes):
        axes[idx].boxplot(df[col].dropna())
        axes[idx].set_title(f'Boxplot of {col}')
        axes[idx].set_ylabel(col)

plt.tight_layout()
plt.show()
```

```python
# Decision: Keep outliers for some columns as they represent valid data points
# For example: capital-gain, capital-loss, hours-per-week may have legitimate extreme values
# We'll only remove extreme outliers from age if any exist beyond our validation

print("Note: Outliers in capital-gain, capital-loss, and hours-per-week are kept")
print("as they represent legitimate extreme values in the census data.")
print(f"\nCurrent dataset shape: {df.shape}")
```

## Step 6: Standardize Categorical Values

```python
# Check unique values in each categorical column
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("Unique values in categorical columns:")
print("-" * 70)
for col in categorical_cols:
    unique_vals = df[col].unique()
    print(f"\n{col} ({len(unique_vals)} unique values):")
    print(unique_vals)
```

## Step 7: Handle Rare Categories

```python
# Group rare categories (less than 1% frequency) into 'Other'
threshold = 0.01

print("Handling rare categories (frequency < 1%):")
print("-" * 70)

for col in categorical_cols:
    # Calculate frequency
    freq = df[col].value_counts(normalize=True)
    rare_categories = freq[freq < threshold].index.tolist()
    
    if len(rare_categories) > 0:
        print(f"\n{col}: {len(rare_categories)} rare categories found")
        print(f"Rare categories: {rare_categories}")
        # Replace rare categories with 'Other'
        df[col] = df[col].apply(lambda x: 'Other' if x in rare_categories else x)
        print(f"Replaced with 'Other'")
    else:
        print(f"\n{col}: No rare categories found")
```

## Step 8: Reset Index and Final Verification

```python
# Reset index after all cleaning operations
df = df.reset_index(drop=True)

print("Final dataset verification:")
print("-" * 70)
print(f"Final shape: {df.shape}")
print(f"\nData types:")
print(df.dtypes)
print(f"\nMissing values: {df.isnull().sum().sum()}")
print(f"\nDuplicate rows: {df.duplicated().sum()}")
```

# STEP 5: Feature Engineering
This step involves:
1. **Encode categorical variables** using One-Hot and Label Encoding
2. **Create education rank scores** based on education level
3. **Create overtime indicator** (hours > 40)
4. **Group rare categories** into 'Other'
5. **Scale numeric features** using StandardScaler

## 5.1 Create Education Rank Scores

```python
# Create education rank scores based on education level hierarchy
education_rank = {
    'Preschool': 1,
    '1st-4th': 2,
    '5th-6th': 3,
    '7th-8th': 4,
    '9th': 5,
    '10th': 6,
    '11th': 7,
    '12th': 8,
    'HS-grad': 9,
    'Some-college': 10,
    'Assoc-voc': 11,
    'Assoc-acdm': 12,
    'Bachelors': 13,
    'Masters': 14,
    'Prof-school': 15,
    'Doctorate': 16
}

df['education_rank'] = df['education'].map(education_rank)

# Handle any unmapped education values (e.g., 'Other' from rare category grouping)
df['education_rank'] = df['education_rank'].fillna(df['education-num'])

print("Education Rank Scores Created:")
print("-" * 50)
print(df[['education', 'education-num', 'education_rank']].drop_duplicates().sort_values('education_rank'))
print(f"\nNew column 'education_rank' added. Shape: {df.shape}")
```

## 5.2 Create Overtime Indicator (hours > 40)

```python
# Create overtime indicator: 1 if hours-per-week > 40, else 0
df['overtime'] = (df['hours-per-week'] > 40).astype(int)

print("Overtime Indicator Created:")
print("-" * 50)
print(f"Workers with overtime (>40 hrs/week): {df['overtime'].sum()} ({df['overtime'].mean()*100:.1f}%)")
print(f"Workers without overtime (<=40 hrs/week): {(df['overtime']==0).sum()} ({(1-df['overtime'].mean())*100:.1f}%)")

# Show overtime vs income relationship
overtime_income = pd.crosstab(df['overtime'], df['income'], normalize='index') * 100
overtime_income.index = ['No Overtime (<=40h)', 'Overtime (>40h)']
print(f"\nOvertime vs Income Distribution:")
print(overtime_income.round(2))
print(f"\nNew column 'overtime' added. Shape: {df.shape}")
```

## 5.3 Group Rare Categories

```python
# Group rare categories (< 2% frequency) into 'Other' for feature engineering
# This reduces dimensionality and avoids sparse one-hot columns

categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
# Exclude 'income' (target variable) from grouping
categorical_cols = [col for col in categorical_cols if col != 'income']

threshold = 0.02
print("Grouping Rare Categories (frequency < 2%):")
print("-" * 70)

for col in categorical_cols:
    freq = df[col].value_counts(normalize=True)
    rare_cats = freq[freq < threshold].index.tolist()
    
    if len(rare_cats) > 0:
        before_unique = df[col].nunique()
        df[col] = df[col].apply(lambda x: 'Other' if x in rare_cats else x)
        after_unique = df[col].nunique()
        print(f"\n{col}:")
        print(f"  Rare categories merged: {rare_cats}")
        print(f"  Unique values: {before_unique} → {after_unique}")
    else:
        print(f"\n{col}: No rare categories (all >= 2%)")

print(f"\nDataset shape after grouping: {df.shape}")
```

## 5.4 Encode Categorical Variables (One-Hot & Label Encoding)

```python
from sklearn.preprocessing import LabelEncoder

# ---- Label Encoding for binary/ordinal columns ----
label_cols = ['sex', 'income']
label_encoders = {}

print("LABEL ENCODING (binary/ordinal columns):")
print("-" * 50)
for col in label_cols:
    le = LabelEncoder()
    df[col + '_encoded'] = le.fit_transform(df[col])
    label_encoders[col] = le
    print(f"\n{col}:")
    mapping = dict(zip(le.classes_, le.transform(le.classes_)))
    print(f"  Mapping: {mapping}")

# ---- One-Hot Encoding for nominal categorical columns ----
onehot_cols = ['workclass', 'marital-status', 'occupation', 'relationship', 'race', 'native-country']

print("\n\nONE-HOT ENCODING (nominal columns):")
print("-" * 50)
for col in onehot_cols:
    print(f"\n{col}: {df[col].nunique()} categories → {df[col].nunique()} new columns")

df_encoded = pd.get_dummies(df, columns=onehot_cols, prefix_sep='_', drop_first=False, dtype=int)

print(f"\n\nDataset shape before encoding: {df.shape}")
print(f"Dataset shape after One-Hot encoding: {df_encoded.shape}")
print(f"New columns added: {df_encoded.shape[1] - df.shape[1]}")

# Show the new columns
new_cols = [c for c in df_encoded.columns if c not in df.columns]
print(f"\nNew encoded columns ({len(new_cols)}):")
for i in range(0, len(new_cols), 5):
    print(f"  {new_cols[i:i+5]}")
```

## 5.5 Scale Numeric Features

```python
from sklearn.preprocessing import StandardScaler

# Select numeric columns to scale (exclude target and encoded binary columns)
numeric_features = ['age', 'fnlwgt', 'education-num', 'education_rank', 
                     'capital-gain', 'capital-loss', 'hours-per-week']

scaler = StandardScaler()

# Create scaled versions with '_scaled' suffix
df_scaled = df_encoded.copy()
df_scaled[numeric_features] = scaler.fit_transform(df_scaled[numeric_features])

print("NUMERIC FEATURE SCALING (StandardScaler):")
print("-" * 70)
print(f"\nScaled features: {numeric_features}")

# Show before vs after comparison
print(f"\n{'Feature':<20} {'Original Mean':>15} {'Original Std':>15} {'Scaled Mean':>15} {'Scaled Std':>12}")
print("-" * 80)
for col in numeric_features:
    orig_mean = df_encoded[col].mean()
    orig_std = df_encoded[col].std()
    scaled_mean = df_scaled[col].mean()
    scaled_std = df_scaled[col].std()
    print(f"{col:<20} {orig_mean:>15.2f} {orig_std:>15.2f} {scaled_mean:>15.4f} {scaled_std:>12.4f}")

print(f"\nFinal engineered dataset shape: {df_scaled.shape}")
print(f"\nColumn list:")
print(df_scaled.columns.tolist())
```

```python
# Step 5 Summary: Feature Engineering Results
print("=" * 70)
print(" " * 15 + "STEP 5: FEATURE ENGINEERING SUMMARY")
print("=" * 70)

print("\n 5.1 Education Rank Scores:")
print(f"     • Created 'education_rank' column mapping education levels to 1-16")

print("\n 5.2 Overtime Indicator:")
print(f"     • Created 'overtime' column (1 if hours > 40, else 0)")
print(f"     • {df['overtime'].mean()*100:.1f}% of workers have overtime")

print("\n 5.3 Rare Category Grouping:")
print(f"     • Categories with < 2% frequency merged into 'Other'")

print("\n 5.4 Categorical Encoding:")
print(f"     • Label Encoded: sex, income (binary columns)")
print(f"     • One-Hot Encoded: workclass, marital-status, occupation, relationship, race, native-country")
print(f"     • Columns expanded: {df.shape[1]} → {df_encoded.shape[1]}")

print("\n 5.5 Numeric Feature Scaling:")
print(f"     • StandardScaler applied to: {numeric_features}")
print(f"     • All numeric features now have mean ≈ 0 and std ≈ 1")

print(f"\n Final dataset shape: {df_scaled.shape}")
print(f" Ready for model training!")
print("=" * 70)
```
