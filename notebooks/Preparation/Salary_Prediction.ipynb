Outliers

```python
print("Identifying ananomalies in numerical columns:")
Q1 = df[numerical_data].quantile(0.25)
Q3 = df[numerical_data].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df[numerical_data] < lower_bound) | (df[numerical_data] > upper_bound)]
outliers.count()
```

```python
# percentage of outliers in each numerical column
outlier_percent = (outliers.count() / len(df)) * 100
print(outlier_percent.round(2).astype(str) + "%")
```

```python
for i in numerical_data:
    sns.boxplot(x=df[i])
    plt.title(f'Outliers for {i}')
    plt.show()
```

#### Cleaning outliers of age, capital-gain, capital-loss, and hours-per-week

```python
# cleaning outliers of age using IQR method
age_Q1 = np.quantile(df['age'], 0.25)
age_Q3 = np.quantile(df['age'], 0.75)
age_IQR = age_Q3 - age_Q1
age_lower_bound = age_Q1 - 1.5 * age_IQR
age_upper_bound = age_Q3 + 1.5 * age_IQR
df = df[(df['age'] >= age_lower_bound) & (df['age'] <= age_upper_bound)]
```

```python
sns.boxplot(x=df['age'])
plt.title('Outliers for age after cleaning')
plt.show()
```

```python
# Drop column fnlwgt as it is not relevant for salary prediction
df.drop(columns=['fnlwgt'], inplace=True)
```

```python
# transform outliers of capital-gain and capital-loss using log
df['capital-gain'] = np.log1p(df['capital-gain'])
df['capital-loss'] = np.log1p(df['capital-loss'])
```

```python
sns.boxplot(x=df['capital-gain'])
plt.title('Outliers for capital-gain after cleaning')
plt.show()
```

```python
sns.boxplot(x=df['capital-loss'])
plt.title('Outliers for capital-gain after cleaning')
plt.show()
```

```python
df['hours-per-week'] = df['hours-per-week'].clip(upper=80)
```

```python
sns.boxplot(x=df['hours-per-week'])
```

Step 4: Statistical Analysis & Insight Extraction

1.Group-by comparisons for income trends

```python
income_count = df['income'].value_counts(normalize=True) * 100
ax = sns.barplot(x=income_count.index, y=income_count.values)
ax.set_xlabel('Income')
ax.set_ylabel('Percentage (%)')
ax.set_title('Income Distribution (%)')
ax.bar_label(ax.containers[0], fmt='%.1f%%', padding=3)
plt.show()
```

```python
pd.crosstab(df['education'], df['income'])
```

```python
pd.crosstab(df['education'], df['income']).plot(kind='bar', stacked=True)
```

```python
edu_group_map = {
    'Preschool': 'Low', '1st-4th': 'Low', '5th-6th': 'Low',
    '7th-8th': 'Low', '9th': 'Low', '10th': 'Low', '11th': 'Low', '12th': 'Low',
    'HS-grad': 'Mid', 'Some-college': 'Mid', 'Assoc-voc': 'Mid', 'Assoc-acdm': 'Mid',
    'Bachelors': 'High', 'Masters': 'High', 'Prof-school': 'High', 'Doctorate': 'High'
}
df['edu_group'] = df['education'].map(edu_group_map)
sns.set_style('whitegrid')
palette = {'<=50K': '#3498db', '>50K': '#e74c3c'}
```

```python
# High Education: Bachelors, Masters, Prof-school, Doctorate
high_df = df[df['edu_group'] == 'High']
ax = sns.countplot(x='education', hue='income', data=high_df,
                   order=['Bachelors', 'Masters', 'Prof-school', 'Doctorate'],
                   palette=palette, edgecolor='white', linewidth=0.8)
for container in ax.containers:
    ax.bar_label(container, fontsize=9, padding=2)
ax.set_title('High Education', fontsize=14, fontweight='bold')
ax.set_xlabel('Education Level', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
plt.xticks(rotation=40, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Mid Education: HS-grad, Some-college, Assoc-voc, Assoc-acdm
mid_df = df[df['edu_group'] == 'Mid']
ax = sns.countplot(x='education', hue='income', data=mid_df,
                   order=['HS-grad', 'Some-college', 'Assoc-voc', 'Assoc-acdm'],
                   palette=palette, edgecolor='white', linewidth=0.8)
for container in ax.containers:
    ax.bar_label(container, fontsize=9, padding=2)
ax.set_title('Mid Education', fontsize=14, fontweight='bold')
ax.set_xlabel('Education Level', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
plt.xticks(rotation=40, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Low Education: Preschool through 12th grade
low_df = df[df['edu_group'] == 'Low']
ax = sns.countplot(x='education', hue='income', data=low_df,
                   order=['Preschool', '1st-4th', '5th-6th', '7th-8th', '9th', '10th', '11th', '12th'],
                   palette=palette, edgecolor='white', linewidth=0.8)
for container in ax.containers:
    ax.bar_label(container, fontsize=9, padding=2)
ax.set_title('Low Education', fontsize=14, fontweight='bold')
ax.set_xlabel('Education Level', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
plt.xticks(rotation=40, ha='right')
plt.tight_layout()
plt.show()
```
